### Remocao Aleatória após Ordenar por ACE

```python
import types
import numpy as np
import pandas as pd
import fac_diagnostico_ace as fac_diag


def selecionar_grupos_aleatorio(
    grupos,
    grupo_ids,
    acertos,
    seed=42,
    jogo=0,
):
    if len(grupos) not in (16, 22):
        raise ValueError("Esperado 16 ou 22 grupos.")

    if len(grupo_ids) != len(grupos) or len(set(grupo_ids)) != len(grupos):
        raise ValueError("GRUPO_ID deve ser único por jogo.")

    for grupo in grupos:
        fac_diag.validar_grupo(grupo, 20, "GRUPO")

    acertos = np.asarray(acertos)

    if (
        acertos.shape != (len(grupos),)
        or not np.issubdtype(acertos.dtype, np.integer)
        or np.any((acertos < 10) | (acertos > 15))
    ):
        raise ValueError("Esperado um ACE inteiro entre 10 e 15 por grupo.")

    # Mesma semente de ordenação usada no diagnóstico por cosseno.
    rng_ordem = np.random.default_rng(
        np.random.SeedSequence([int(seed), int(jogo), 1])
    )

    # Sorteios de remoção independentes da ordenação.
    rng_remocao = np.random.default_rng(
        np.random.SeedSequence([int(seed), int(jogo), 2])
    )

    ativos = np.arange(len(grupos))
    auditoria = []
    rodada = 0

    while len(ativos) > 6:
        rodada += 1

        embaralhados = rng_ordem.permutation(ativos)
        ordem = embaralhados[
            np.argsort(-acertos[embaralhados], kind="stable")
        ]

        n_blocos = len(ordem) // 4
        limite = n_blocos * 4
        remover = set()

        for bloco in range(n_blocos):
            indices = ordem[4 * bloco:4 * bloco + 4]

            # Cada um dos quatro grupos tem 25% de chance de sair.
            excluido = int(rng_remocao.choice(indices))
            remover.add(excluido)

            for indice in indices:
                auditoria.append({
                    "RODADA": rodada,
                    "N_ANTES": len(ativos),
                    "N_DEPOIS": len(ativos) - n_blocos,
                    "BLOCO": bloco + 1,
                    "GRUPO_ID": grupo_ids[indice],
                    "PROB_REMOCAO": 0.25,
                    "REMOVIDO": int(indice) == excluido,
                    "SOBRA": False,
                })

        for indice in ordem[limite:]:
            auditoria.append({
                "RODADA": rodada,
                "N_ANTES": len(ativos),
                "N_DEPOIS": len(ativos) - n_blocos,
                "BLOCO": 0,
                "GRUPO_ID": grupo_ids[indice],
                "PROB_REMOCAO": 0.0,
                "REMOVIDO": False,
                "SOBRA": True,
            })

        ativos = np.array([
            i for i in ativos if int(i) not in remover
        ])

    return ativos, pd.DataFrame(auditoria)


# Reutiliza a avaliação anterior com o novo seletor, sem alterar
# o módulo nem interferir no teste por cosseno em andamento.
contexto_aleatorio = dict(fac_diag.executar_experimento.__globals__)
contexto_aleatorio["selecionar_grupos"] = selecionar_grupos_aleatorio

executar_aleatorio = types.FunctionType(
    fac_diag.executar_experimento.__code__,
    contexto_aleatorio,
    name="executar_aleatorio",
    argdefs=fac_diag.executar_experimento.__defaults__,
)


# Ordenação por ACE + remoção aleatória: 100 sementes.
linhas = []

for seed in range(100):
    resultado = executar_aleatorio(df, seed=seed)

    linhas.append(
        resultado["resumo"]
        .set_index("METRICA")["VALOR"]
        .to_dict()
    )

    if (seed + 1) % 10 == 0:
        print(f"Sementes concluídas: {seed + 1}/100", flush=True)

repeticoes_ace_aleatorio = pd.DataFrame(linhas)

display(
    repeticoes_ace_aleatorio[
        ["HIT6_CONDICIONAL_PCT", "GANHO_CONDICIONAL_PP"]
    ]
    .agg(["mean", "std", "min", "max"])
    .round(2)
)
```
